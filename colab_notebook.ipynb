```python


import os
import fitz          
import pdfplumber

try:
    import camelot
    HAS_CAMELOT = True
except Exception as e:
    HAS_CAMELOT = False
    print(f"NOTE: camelot unavailable ({e}). Skipping Camelot tests.\n")



FILES_TO_TEST = [
    r"raw data\olive\olive leaf spot.pdf",
    r"raw data\citrus\leaf spot\Alternaria_diseases_of_citrus_-_Novel_pathosystems.pdf",
    r"raw data\liste-des-produits-pesticides-homologues-16-juillet-2025-vf-1.pdf",
]


def test_pymupdf(file_path):
    print("--- PyMuPDF (plain text) ---")
    doc = fitz.open(file_path)
    for i, page in enumerate(doc):
        text = page.get_text("text")
        print(f"[page {i}] {len(text)} chars")
        print(text[:500])
        print("...")
    doc.close()


def test_pdfplumber(file_path):
    print("\n--- pdfplumber (table detection) ---")
    with pdfplumber.open(file_path) as pdf:
        for i, page in enumerate(pdf.pages):
            tables = page.find_tables()
            print(f"[page {i}] {len(tables)} table(s) detected")
            if tables:
                extracted = page.extract_tables()
                for t_idx, table in enumerate(extracted):
                    print(f"  table {t_idx}: {len(table)} rows x "
                          f"{len(table[0]) if table else 0} cols")
                    for row in table[:2]:
                        print("   ", row)


def test_camelot(file_path):
    if not HAS_CAMELOT:
        return
    print("\n--- Camelot (lattice mode) ---")
    try:
        tables = camelot.read_pdf(file_path, pages="all", flavor="lattice")
        print(f"{len(tables)} table(s) found across all pages")
        for i, t in enumerate(tables):
            print(f"  table {i}: accuracy={t.parsing_report['accuracy']:.1f}%, "
                  f"shape={t.df.shape}")
    except Exception as e:
        print(f"Camelot failed on this file: {e}")


# ============================================================
# RUN — this executes immediately, no __main__ guard, no argv.
# ============================================================
for file_path in FILES_TO_TEST:
    print("=" * 70)
    print(f"FILE: {file_path}")
    print("=" * 70)

    if not os.path.exists(file_path):
        print(f"FILE NOT FOUND: {file_path}")
        print("Check the exact spelling/spacing/accents against your real folder.\n")
        continue

    test_pymupdf(file_path)
    test_pdfplumber(file_path)
    test_camelot(file_path)
    print("\n")

print("Done. Review the output above for each file:")
print("- PyMuPDF: does the text look clean and in the right order?")
print("- pdfplumber: did it find tables where you expect them (e.g. pesticide catalog)?")
print("- Camelot: high accuracy (>90%) = trustworthy, low/skipped = fall back to pdfplumber.")
```
