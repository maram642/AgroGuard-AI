```python
import os
import pandas as pd


# ==============================
# Configuration
# ==============================

# Folder containing your PDFs
BASE_FOLDER = "raw data"

# Output file
OUTPUT_FILE = "metadata.xlsx"


# ==============================
# Scan PDFs
# ==============================

metadata = []

for root, dirs, files in os.walk(BASE_FOLDER):

    for file in files:

        # Only PDFs
        if file.lower().endswith(".pdf"):

            # Full relative path
            file_path = os.path.join(root, file)

            # Split path
            path_parts = file_path.split(os.sep)

            # Get crop name
            # Example:
            # raw_data / olive / disease.pdf
            crop = path_parts[-2]


            # Disease name from filename
            disease = os.path.splitext(file)[0]


            metadata.append({

                "file_path": file_path.replace("\\", "/"),

                "crop": crop,

                "disease": disease,

                "agent": "",

                "document_type": "disease",

                "country": "",

                "region": "",

                "source": "",

                "year": ""

            })


# ==============================
# Create dataframe
# ==============================

df = pd.DataFrame(metadata)


# Sort by crop
df = df.sort_values(
    by=["crop", "disease"]
)


# Save Excel
df.to_excel(
    OUTPUT_FILE,
    index=False
)


print("Metadata created successfully!")
print(f"Total PDFs found: {len(df)}")
print(f"Saved as: {OUTPUT_FILE}")
```
